```python
import pandas as pd
import numpy as np
```

```python
ola=pd.read_excel("Bengaluru Ola.xlsx")
ola.head(10)
```

```python
ola.columns
```

```python
uber=pd.read_excel("Bengaluru Uber.xlsx")
uber.head(10)
```

```python
uber.columns
```

```python
uber.rename(columns={"Customer Ratings":"Customer Rating"},inplace=True)
```

```python
ola["Company"]="Ola"
uber["Company"]="Uber"
```

```python
common_cols=['Date', 'Time', 'Booking ID', 'Booking Status', 'Customer ID',
       'Vehicle Type', 'Pickup Location', 'Drop Location', 'Avg VTAT',
       'Avg CTAT','Booking Value', 'Payment Method',
       'Ride Distance', 'Driver Ratings', 'Customer Rating', 'Company']
```

```python
ola=ola[common_cols]
uber=uber[common_cols]
```

```python
rides=pd.concat([ola,uber],ignore_index=True)
```

```python
rides.shape
```

```python
rides["Company"].value_counts()
```

```python
rides.isnull().sum()
```

```python
missing=rides.isnull().sum()
missing_percent=(missing/len(rides))*100
print(missing_percent.sort_values(ascending=False))
```

```python
rides["Booking Status"].value_counts(dropna=False)
```

```python
rides.groupby("Booking Status").agg({
    "Booking Value":"count",
    "Ride Distance":"count",
    "Customer Rating":"count"
})
```

```python
rides["Booking Status"].value_counts()
```

```python
def ride_outcome(status):
    if status=="Success":
        return "Success"
    elif "Cancelled" in status:
        return "Cancelled"
    else :
        return "Incomplete"

rides["Ride_outcome"]=(rides["Booking Status"].apply(ride_outcome))
```

```python
rides["Ride_outcome"].value_counts()
```

```python
pd.crosstab(rides["Company"],rides["Ride_outcome"],normalize="index")*100
```

```python
pricing_df=rides.dropna(
    subset=["Booking Value","Ride Distance"]
).copy()
pricing_df["Fare_per_KM"]=(pricing_df["Booking Value"]/pricing_df["Ride Distance"])
```

```python
pricing_df.head()
```

```python
print(type(rides["Time"].iloc[0]))
```

```python
rides["Hour"]=rides["Time"].apply(lambda x:x.hour)
```

```python
rides[["Time","Hour"]].head()
```

```python
def bucket(hour):
    if 5<=hour<12:
        return "Morning"
    elif 12<=hour<17:
        return "Afternoon"
    elif 17<=hour<22:
        return "Evening"
    else:
        return"Night"

rides["Time_bucket"]=rides["Hour"].apply(bucket)
```

```python
rides["Time_bucket"].value_counts()
```

```python
rides["Date"]=pd.to_datetime(rides["Date"])
rides["Day_of_Week"]=(rides["Date"].dt.day_name())
```

```python
rides["Day_of_Week"].value_counts()
```

```python
rides["Day_Type"]=np.where(rides["Day_of_Week"].isin(["Saturday","Sunday"]),"Weekend","Weekday")
```

```python
pricing_df["Distance_Category"]=pd.cut(pricing_df["Ride Distance"],bins=[0,5,10,20,1000],labels=["Short","Medium","Long","Very Long"])
```

```python
pricing_df["Revenue_Category"]=pd.cut(pricing_df["Booking Value"],bins=[0,200,500,1000,10000],labels=["Low","Medium","High","Premium"])
```

```python
rating_df=rides.dropna(subset=["Customer Rating","Driver Ratings"]).copy()
rating_df["Rating_Difference"]=(rating_df["Customer Rating"]-rating_df["Driver Ratings"])
```

```python
pricing_df.groupby("Company")["Fare_per_KM"].describe()
```

```python
pd.crosstab(rides["Company"],rides["Ride_outcome"],normalize="index")*100
```

```python
vehicle_map = {
    "Bike":"Bike",
    "eBike":"Bike",
    "Auto":"Auto",
    "Mini":"Economy",
    "Prime Sedan":"Sedan",
    "Prime SUV":"SUV",
    "Prime Plus":"Premium",
    "Moto":"Bike",
    "Uber Go":"Economy",
    "Go Sedan":"Sedan",
    "UberXL":"SUV",
    "Premier":"Premium",
    "Non-AC Taxi":"Economy",
    "eBike":"Bike"
}
```

```python
rides["Vehicle_Category"] = (rides["Vehicle Type"].map(vehicle_map))
```
